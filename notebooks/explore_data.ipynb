```python
import os
import matplotlib.pyplot as plt
from PIL import Image
import random
```

```python
from torchvision.datasets import ImageFolder

dataset = ImageFolder("../dataset/train")

print("Classes:", len(dataset.classes))
print("Class names:", dataset.classes)
print("Total images:", len(dataset))
```

```python
from collections import Counter
counts = Counter(dataset.targets)

for idx, class_name in enumerate(dataset.classes):
    print(f"{class_name}: {counts[idx]} images")
```

```python
dataset_path = "../dataset/train"

# Get all class folders
classes = [
    c for c in os.listdir(dataset_path)
    if os.path.isdir(os.path.join(dataset_path, c))
]

# Pick 10 random classes
sample_classes = random.sample(classes, 10)

fig, axes = plt.subplots(2, 5, figsize=(15, 6))
fig.suptitle("Sample Card Images", fontsize=14)

for ax, class_name in zip(axes.flatten(), sample_classes):

    class_path = os.path.join(dataset_path, class_name)

    img_name = random.choice(os.listdir(class_path))
    img_path = os.path.join(class_path, img_name)

    img = Image.open(img_path)

    ax.imshow(img)
    ax.set_title(class_name)
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
cols = 8
rows = (len(classes) + cols - 1) // cols

fig, axes = plt.subplots(rows, cols, figsize=(16, rows * 2))

for ax, class_name in zip(axes.flatten(), classes):

    class_path = os.path.join(dataset_path, class_name)

    img_name = random.choice(os.listdir(class_path))
    img = Image.open(os.path.join(class_path, img_name))

    ax.imshow(img)
    ax.set_title(class_name, fontsize=8)
    ax.axis("off")

# Hide empty plots
for ax in axes.flatten()[len(classes):]:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
sizes = []

for class_name in os.listdir(dataset_path):
    class_path = os.path.join(dataset_path, class_name)

    if os.path.isdir(class_path):

        for fname in os.listdir(class_path)[:10]:  # sample 10 images per class
            img_path = os.path.join(class_path, fname)

            try:
                img = Image.open(img_path)
                sizes.append(img.size)
            except:
                print(f"Could not open {img_path}")

unique_sizes = set(sizes)

print(f"Number of unique image sizes: {len(unique_sizes)}")
print(unique_sizes)
```
